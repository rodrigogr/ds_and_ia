# Clusterização ou Grupamento
Um algoritmo de clusterização é uma técnica de aprendizado de máquina e mineração de dados que agrupa um conjunto de dados em clusters ou grupos com base em suas similaridades. Esses algoritmos são usados para identificar padrões nos dados e organizar as informações em grupos significativos. Alguns exemplos de uso incluem <font color=lightgreen><u>segmentação de mercado</u>, de clientes, análise de redes sociais, coompreensão de imagens e agrupamento de documentos.</font> <u>Essas aplicações permitem direcionar estratégias de marketing, personalizar conteúdo e facilitar a organização e recuperação de informações.</u>

# Medidas de distância

- <font color=lightblue>Distância Euclidiana:</font> É a distância mais comum e simples. Calculada como a raiz quadrada da soma dos quadrados das diferenças entre as coordenadas dos pontos em cada dimensão.
- <font color=lightblue>Distância Manhattan:</font> Também conhecida como distância de cidade, é a soma das diferenças absolutas entre as coordenadas dos pontos em cada dimensão.
- <font color=lightblue>Distância de Minkowski:</font> É uma generalização das distâncias Euclidiana e Manhattan. Dependendo do valor de um parâmetro p, pode reduzir-se à distância Euclidiana (quando p = 2) ou à distância Manhattan (quando p = 1).
- <font color=lightblue>Distância de Chebyshev:</font> É a maior diferença absoluta entre as coordenadas dos pontos em qualquer dimensão.
- <font color=lightblue>Distância de Coseno:</font> É uma medida da similaridade entre dois vetores, definida como o cosseno do ângulo entre eles. É útil para dados em espaços de alta dimensão.
- <font color=lightblue>Distância de Hamming:</font> É usada para dados categóricos e conta o número de dimensões nas quais os pontos diferem.

# K-Means
O K-Means é um algoritmo de aprendizado não supervisionado amplamente utilizado para agrupamento de dados (clustering). Ele <font color lightgreen>organiza pontos de dados não rotulados em grupos (clusters) com base na similaridade</font>, buscando minimizar a variância dentro de cada cluster. Seu objetivo é minimizar a distância intra-cluster, formando agrupamentos coesos e distintos entre si.

Ele <font color lightgreen>agrupa dados em k-clusters</font>, onde k é definido pelos cientistas de dados. O algoritmo começa definindo centroides aleatórios para cada cluster e, em seguida, atribui pontos a cada cluster com base em uma medida de distância, geralmente a distância euclidiana. Os centróides são recalculados como a média dos pontos atribuídos a cada cluster, e esse processo de atribuição e recálculo é repetido até que não haja mudanças significativas ou um número máximo de iterações seja alcançado. O resultado final pode variar dependendo da inicialização aleatória dos centroides e do número de clusters escolhidos.

## Funcionamento do K-Means
O K-Means opera iterativamente em dois passos principais, conhecidos como Etapas de Atribuição e Atualização . O processo é repetido até que os centroides não mudem significativamente ou até que um número máximo de iterações seja atingido.



- <font color=lightgreen>1. Inicialização:</font> Escolha o número de clusters (K), definido previamente pelo usuário, onde representa a quantidade de grupos desejados.
- <font color=lightgreen>2. Inicialização aleatória de centróides:</font> K pontos são selecionados aleatoriamente como centros iniciais dos clusters.
- <font color=lightgreen>3. Atribuição de pontos aos clusters:</font> Cada dado é associado ao centróide mais próximo usando distância euclidiana.
- <font color=lightgreen>4. Atualização dos centróides:</font> Recalcula-se a posição dos centros como média dos pontos atribuídos a cada cluster.
- <font color=lightgreen>5. Repetição:</font> Repita as etapas 3-Atribuição e 4-Atualização até a convergência (centroides não mudem significativamente), ou seja, os centróides se estabilizam.

## Características principais
- <font color=yellow>Não determinístico:</font> Resultados podem variar devido à inicialização aleatória.
- <font color=yellow>Métrica de avaliação:</font> Utiliza inércia (soma das distâncias quadradas entre pontos e seus centróides) para otimização.
- <font color=yellow>Escolha do K ideal:</font> Técnicas como <u>método do cotovelo, Silhouette Score e Gap Statistic</u>.

## Vantagens do K-Means

- <font color=middleblue>Simplicidade:</font> O algoritmo é fácil de entender e implementar.
- <font color=middleblue>Eficiência:</font> É computacionalmente eficiente para grandes conjuntos de dados, especialmente quando otimizado.
- <font color=middleblue>Escalabilidade:</font> Funciona bem com grandes volumes de dados.

## Desvantagens do K-Means

- <font color=red>Número de Clusters Fixo:</font> O valor de K deve ser especificado previamente, o que pode ser problemático se o número ideal de clusters for desconhecido.
- <font color=red>Sensibilidade à Inicialização:</font> A escolha inicial dos centroides pode influenciar significativamente o resultado final.
- <font color=red>Assume Forma Esférica:</font> O K-Means tende a criar clusters de formato esférico e tamanho semelhante, o que pode ser inadequado para dados com formas irregulares.
- <font color=red>Sensível a Outliers:</font> Pontos de dados muito distantes podem distorcer os centroides.

## Métodos para Determinar o Número Ideal de Clusters (K)
Existem diferentes métodos para definir o número ideal de clusters em um algoritmo de caminhos. O <font color=lightblue>método do cotovelo</font> envolve plotar a função de custo em relação ao número de clusters e observar quando ocorre uma mudança significativa na inclinação da curva. O <font color=lightblue>método da silhueta</font> avalia a qualidade dos clusters formados por diferentes valores de K e escolhe o valor que maximiza a média da silhueta. O <font color=lightblue>método Gap Statistics</font> compara a dispersão intra-cluster para diferentes valores de K e escolhe o número ideal de clusters que maximiza a lacuna estatística.

<font color=lightblue>Método do Cotovelo (Elbow Method):</font>

- Plote a função de custo (J) em relação ao número de clusters (K).
- Escolha o valor de K onde a redução na função de custo começa a diminuir significativamente (formando um "cotovelo").

<font color=lightblue>Silhouette Score:</font>
- Mede quão bem um ponto de dado se ajusta ao seu próprio cluster em comparação com outros clusters.
- Valores mais altos indicam melhor clusterização.

<font color=lightblue>Gap Statistic:</font>
- Compara a inércia do clustering real com a inércia esperada sob uma distribuição nula (dados aleatórios).

<font color=lightblue>Validação Externa:</font> 
- Em alguns casos, pode-se ter acesso a informações externas sobre os dados que indicam o número correto de clusters. Nesses casos, métodos de validação externa, como índices de validação externa ou comparação com rótulos conhecidos, podem ser usados para determinar o número ideal de clusters.

<font color=lightblue>Conhecimento de Domínio:</font> 
- Às vezes, o conhecimento prévio do domínio pode ajudar a determinar o número de clusters de forma mais precisa. Por exemplo, em ciências sociais, pode-se ter conhecimento sobre o número de grupos ou categorias existentes.

Para testar essa hipótese, utilizaremos o teste estatístico <font color=lightblue>ANOVA (Análise de Variância)</font>, que verifica se há variações significativas nas médias de diferentes grupos. No entanto, <u>antes de aplicar o teste, precisamos validar algumas suposições, como a independência das observações, a normalidade dos dados e a homogeneidade das variâncias.</u> Caso essas suposições não sejam atendidas, utilizaremos o <font color=lightblue>teste de ANOVA de Welch</font>, que é mais robusto nessas situações.

```python
# Distribuição da variável Inovação
percentual_inovacao = df_clientes.value_counts('inovacao')/len(df_clientes)*100
px.bar(percentual_inovacao, color=percentual_inovacao.index)
```

## Teste ANOVA (Análise de Variância)
<font color=orange>Analisar os pressupostos necessários para realizar um teste estatístico de ANOVA.</font>
- Observações precisam ser independentes umas das outras.
- Variável dependente precisa ser contínia.
- Verificar se os dados da variável dependente seguem uma distribuição normal utilizando o <font color=lightgreen>teste de Shapiro-Wilk</font>.
- Verificar se as variâncias entre os grupos são homogêneas utilizando o <font color=lightblue>teste de Bartlett</font>.
- As amostras devem tem tamanhos iguais.

### Teste estatístico de Bartlett
Verificar se as variâncias entre os grupos são homogêneas

```python
# Checar se as variâncias (faturamento) entre os grupos (inovação) são homogêneas
# Aplicar Teste de Bartlett
# H0 - Variâncias são iguais
# H1 - Variâncias não são iguais

# Separando os dados de faturamento com base na coluna 'inovacao'
dados_agrupados = [df_clientes['faturamento_mensal'][df_clientes['inovacao'] == grupo] for grupo in df_clientes['inovacao'].unique()] # For em todos possíveis valores de inovação (0 a 9)

# Executar o teste de Bartlett
bartlett_test_statistic, bartlett_p_value = bartlett(*dados_agrupados)

# Exibindo os resultados
print(f'Estatística do Teste de Bartlett: {bartlett_test_statistic}')
print(f'P-Value do Teste de Bartlett: {bartlett_p_value}')
print('Não rejeitamos H0, portanto as variâncias são iguais')
```

### Teste de Shapiro-Wilk
Verificar se os dados da variável dependente seguem uma distribuição normal

```python
# Checar se os dados seguem uma distribuição normal
# Aplicar Teste de Shapiro-Wilk
# H0 - Segue uma distribuição normal
# H1 - Não segue uma distribuição normal

# Executar o teste
shapiro_test_statistic, shapiro_p_value = shapiro(df_clientes['faturamento_mensal'])

# Exibindo os resultados
print(f'Estatística do Teste de Shapiro-Wilk: {shapiro_test_statistic}')
print(f'P-Value do Teste de Shapiro-Wilk: {shapiro_p_value}')
print('Não rejeitamos H0, portanto os valores de faturamento mensal segue uma distribuição normal')
```

## ANOVA de Welch
Utilizamos o teste de ANOVA de Welch, pois as amostras têm tamanhos diferentes.


```python
# H0 - Não há diferenças significativas entre as médias dos grupos
# H1 - Há pelo menos uma diferença significativa entre as médias dos grupos

aov = welch_anova(dv='faturamento_mensal', between='inovacao', data=df_clientes)

# Exibindo Resultados
print(f'Estatística do Teste de ANOVA Welch: {aov.loc[0,'F']}')
print(f'P-Value do Teste de ANOVA Welch: {aov.loc[0,'p-unc']}')
print('O resultado do teste mostrou que não há diferenças significativas entre as médias dos grupos, independentemente do nível de inovação das empresas.')
print('Isso significa que empresas com diferentes níveis de inovação podem ter faturamentos semelhantes.')
print('Não rejeitamos H0, portanto Não há diferenças significativas entre as médias dos grupos.') 
```

# Treinamento


## Preparando e treinando os dados para o KMeans
Para treinar o algoritmo, precisamos <u>preparar os dados</u>.
- Selecionaremos todas as colunas, pois não temos um target específico. 
- Separaremos as variáveis numéricas, categóricas e ordinais. 
- Aplicaremos transformações em cada tipo de variável usando um <u>standard scaler, one-hot-encoder e ordinal encoder</u>. 
- Criaremos um <u>pré-processador</u> que transformará os dados em um novo dataset. 
- Usaremos o <u>column transformer</u> para indicar quais transformadores usar. 
- Faremos o <u>fit transform no pré-processador</u> para treinar os transformadores e aplicar as transformações nos dados.

```python
# Selecionar as colunas para clusterização
X = df_clientes.copy()

# Separar as variáveis numéricas categóricas nominais e ordinais
numeric_features = ['faturamento_mensal', 'numero_de_funcionarios', 'idade']
categorical_features = ['localizacao', 'atividade_economica']
ordinal_features = ['inovacao']

# Aplicar Transformações por tipo
numeric_transformer = StandardScaler()
categorical_transformer = OrdinalEncoder()
ordinal_transformer = OrdinalEncoder()

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features),
        ('ord', ordinal_transformer, ordinal_features)
    ]
)

# Tranformar os dados
X_trasnformed = preprocessor.fit_transform(X)
```

# Métricas de Clusterização
- <font color=orange>Índice de Silhueta:</font> Avalia a coesão intra-cluster e a separação inter-cluster. Uma pontuação alta indica que os pontos estão bem agrupados dentro de seus clusters e mal conectados aos clusters vizinhos.
- <font color=orange>Índice Davies-Bouldin (DB):</font> Mede a dispersão dentro de cada cluster em relação à separação entre clusters. Quanto menor o valor, melhor a zeparação entre os clusters.
- <font color=orange>Índice de Calinski-Harabasz (CH):</font> Calcula a relação entre a dispersão intra-cluster e a dispersão entre clusters. Pontua mais alto para clusters densos e bem separados.
- <font color=orange>Índice Dunn:</font> É a razão entre a menor distância inter-cluster e a maior distância intra-cluster. Pontua mais alto para clusters mais compactos e mais distantes uns dos outros.
- <font color=orange>Índice Rand Adjustado (ARI):</font> Comparação entre os rótulos atribuídos pelos algoritmos de clusterização e os rótulos verdadeiros, quando disponíveis.
- <font color=orange>Índice de Validade Interna (IV):</font> Consiste em várias medidas internas, como compacidade e separação dos clusters, para avaliar a qualidade da clusterização sem rótulos verdadeiros.



##  Métrica para ajuste de hiperparâmetros
 
- Com o Optuna iremos <u>otimizar a métrica de silhueta para ajustar os hiperparâmetros</u>. 
- Definiremos uma função chamada <u>K-Means Objective, que receberá uma tentativa de ajuste dos hiperparâmetros</u>. 
- Ajustaremos dois hiperparâmetros: o <u>número de clusters (K) e a distância (Euclidiana e Minkowski)</u>. 
- Instanciaremos o K-Means com esses hiperparâmetros e treinaremos o modelo. 
- Calcularemos o <u>Silhouette Score usando a função pairwise distance e os rótulos do K-Means</u>. 
- Criaremos um estudo do Optuna e rodaremos o experimento para encontrar a melhor configuração de hiperparâmetros.

```python
# Optuna para Otimização de Hiperparâmetros
def kmeans_objective(trial):

    # Definindo os hiperparâmetros a serem ajustados
    n_clusters = trial.suggest_int('n_clusters', 3, 10)
    distance_metric = trial.suggest_categorical('distance_metric', ['euclidean', 'minkowski'])

    # Criar modelo
    modelo_kmeans = KMeans(n_clusters=n_clusters, random_state=51)

    # Treinar Modelo
    modelo_kmeans.fit(X_trasnformed)

    # Calculando o Silhouette Score
    distances = pairwise_distances(X_trasnformed, metric=distance_metric)
    silhouette_avg = silhouette_score(distances, modelo_kmeans.labels_)

    return silhouette_avg

# Criar estudo do Optuna
search_space = {'n_clusters': [3, 4, 5, 6, 7, 8, 9, 10], 'distance_metric': ['euclidean', 'minkowski']}
sampler = optuna.samplers.GridSampler(search_space=search_space)
estudo_kmeans = optuna.create_study(direction='maximize', sampler=sampler)

# Rodar Estudo
estudo_kmeans.optimize(kmeans_objective, n_trials=100)
```

# Análise de Resultados

- Vamos <u>analisar os resultados do Optuna e treinar um modelo com os melhores parâmetros encontrados</u>.
- Guardaremos a melhor configuração em uma variável chamada <u>best_params</u>. 
- Criaremos o modelo usando esses melhores parâmetros e ajustaremos aos dados. 
- Calcularemos o score de silhueta para avaliar o desempenho do modelo. 
- Por fim, atribuiremos os clusters aos registros do nosso dataset e visualizar os primeiros registros com a coluna de cluster atualizada.

```python
# Melhor configuração encontrada pelo Optuna
best_params = estudo_kmeans.best_params

# Instanciando o modelo K-means com melhores parâmetros
best_kmeans = KMeans(n_clusters=best_params['n_clusters'], random_state=51)
best_kmeans.fit(X_trasnformed)

# Calculando o Silhouette Score
distances = pairwise_distances(X_trasnformed, metric=best_params['distance_metric'])
best_silhouette = silhouette_score(distances, best_kmeans.labels_)

# Exibindo Resultados
print(f'k (Número de Clusters): {best_params['n_clusters']}')
print(f'Métrica de Distância selecionada: {best_params['distance_metric']}')
print(f'Silhouette Score: {best_silhouette}')
```

```python
# Criar coluna com cluster escolhido
df_clientes['cluster'] = best_kmeans.labels_
df_clientes.head()
```

## Entrega do Modelo com APP Batch "Gradio"
- Desenvolvemos uma aplicação simples no Gradio. 
- Utilizamos um algoritmo de clusterização não supervisionado para transformar os dados em um pipeline separado. 
- Não fizemos predições, apenas trouxemos uma estrutura com o algoritmo.
  
<font color=lightblue>Carregamos o modelo e definimos uma função chamada "clustering" que recebe um arquivo CSV, onde:</font>
1. Lê o arquivo 
2. Transforma os dados para o formato necessário
3. Treina o modelo 
4. Adiciona a coluna de cluster no DataFrame 
5. Salvamos o DataFrame em um arquivo CSV e retornamos o nome desse arquivo. 
6. Criamos uma interface para chamar a função e rodamos a aplicação. 
7. Com isso, podemos fazer upload de um arquivo, que será processado e gerará um arquivo de saída com os clusters.

```python
modelo = joblib.load('./modelo_clusterizacao_clientes.pkl')
preprocessor = joblib.load('./pipeline_clusterizacao_clientes.pkl')

def clustering(arquivo):
    # 1º Lê o arquivo
    df_empresas = pd.read_csv(arquivo.name)
    
    # 2º Transformar os dados do DF para o formato que o Kmeans precisa
    X_trasnformed = preprocessor.fit_transform(df_empresas)
    
    # 3º Treinar o modelo
    modelo.fit(X_trasnformed)
    
    #4º Criar a coluna cluster no DF
    df_empresas['Cluster'] = modelo.labels_

    # 5º Salva o DataFrame em um arquivo CSV e retorna o nome desse arquivo
    df_empresas.to_csv('./clusters.csv', index=False)    
    return './clusters.csv'
```

# Criar a interface Batch com Gradio
Criamos uma interface batch com Gradio para chamar a função e rodamos a aplicação.

```python
# 6º Criar a interface e rodar a aplicação
app = gr.Interface(
    clustering,
    gr.File(file_types=[".csv"]),
    "file"
)
app.launch()
```
